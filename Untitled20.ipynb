### Act 3 listas con funciones

```python
def listaringenieros(lista, titulo="Lista de Ingenieros"):
    """Muestra la lista actual de ingenieros."""
    print(f"\n--- {titulo} ---")
    if not lista:
        print("La lista está vacía.")
    else:
        for i, ing in enumerate(lista, 1):
            print(f"{i}. {ing}")
    print("-" * (len(titulo) + 8))

def leeringeniero():
    """Solicita y retorna el nombre de un ingeniero."""
    return input("Ingrese el nombre del ingeniero: ").strip()

def agregaringeniero(lista):
    """Añade un nuevo ingeniero a la lista."""
    listaringenieros(lista, "ESTADO ANTES DE AGREGAR")
    nuevo = leeringeniero()
    lista.append(nuevo)
    print(f"\n✅ '{nuevo}' ha sido agregado.")
    listaringenieros(lista, "ESTADO DESPUÉS DE AGREGAR")

def buscaringeniero(lista):
    """Busca un ingeniero y retorna su índice si existe."""
    nombre = leeringeniero()
    if nombre in lista:
        indice = lista.index(nombre)
        print(f"🔍 Encontrado: '{nombre}' en la posición {indice + 1}.")
        return indice
    else:
        print(f"❌ El ingeniero '{nombre}' no se encuentra en la lista.")
        return None

def modificaringeniero(lista):
    """Modifica un nombre existente en la lista."""
    listaringenieros(lista, "ESTADO ANTES DE MODIFICAR")
    indice = buscaringeniero(lista)

    if indice is not None:
        nuevo_nombre = input("Ingrese el nuevo nombre: ").strip()
        lista[indice] = nuevo_nombre
        print("\n📝 Cambio realizado con éxito.")
        listaringenieros(lista, "ESTADO DESPUÉS DE MODIFICAR")

def borraringeniero(lista):
    """Elimina un ingeniero de la lista."""
    listaringenieros(lista, "ESTADO ANTES DE BORRAR")
    indice = buscaringeniero(lista)

    if indice is not None:
        eliminado = lista.pop(indice)
        print(f"\n🗑️ '{eliminado}' ha sido eliminado.")
        listaringenieros(lista, "ESTADO DESPUÉS DE BORRAR")

def menu():
    # Lista inicial con 5 ingenieros
    ingenieros = ["Elon Musk", "Margaret Hamilton", "Steve Wozniak", "Ada Lovelace", "Nikola Tesla"]

    while True:
        print("\n--- MENÚ DE GESTIÓN DE INGENIEROS ---")
        print("1. Listar Ingenieros")
        print("2. Agregar Ingeniero")
        print("3. Buscar Ingeniero")
        print("4. Modificar Ingeniero")
        print("5. Borrar Ingeniero")
        print("6. Salir")

        opcion = input("\nSeleccione una opción: ")

        if opcion == "1":
            listaringenieros(ingenieros)
        elif opcion == "2":
            agregaringeniero(ingenieros)
        elif opcion == "3":
            buscaringeniero(ingenieros)
        elif opcion == "4":
            modificaringeniero(ingenieros)
        elif opcion == "5":
            borraringeniero(ingenieros)
        elif opcion == "6":
            print("Saliendo del sistema... ¡Ten un excelente día!")
            break
        else:
            print("Opción no válida, intenta de nuevo.")

# Ejecutar el programa
if __name__ == "__main__":
    menu()
print("programa realizado por David Barrera 1388")
```
